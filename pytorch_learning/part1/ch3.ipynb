## Tensors: A Multidimentional Array at its core

#### Lists and Tensors:

```python
l = [1.0, 2.0, 1.0] # Python List --> ZERO bassed indexing
print(l[2])
print(l)
```

```python
import torch
t = torch.ones(3) # 1D Tensor of size 3, filled with ones: A Row of 1's
print(t)
print(t[1])
print(float(t[2]))
t[2] = 2.0 # Assigning value to tensor at idx 2
t
```

Representing a coord system with pytorch 1D tensors (Instead of python lists): \
 --> Xs in even indices, Ys in odd indices (Tensors are contiguous, just like c arrays -> which they are bassed on)

```python
points = torch.zeros(6) # Use .zeros to get an appropriate-sized array, then overwrite 0's with desired values
points[0] = 4.0
points[1] = 1.0 
points[2] = 5.0
points[3] = 3.0
points[4] = 2.0
points[5] = 1.0
print(points)

# We can also pass a python list into the tensor constructor to do the same thing
points_v2 = torch.tensor([4.0, 1.0, 5.0, 3.0, 2.0, 1.0])
print(points_v2)

# Get coords of first point
float(points[0]), float(points[1])
```

Using 2D tensors --> Index represents whole 2D point 

```python
points = torch.tensor([[4.0, 1.0], [5.0, 3.0], [2.0, 1.0]])
print(points)
print(points[1]) # Output is another tensor **that is just a different view of the underlying data**
print(points.shape, '\n') # Get shape of tensor (size/dimensions)

# Initializing a tensor (of zeros or ones) by providing the shape as a tuple
points = torch.zeros(3, 2)
print(points, '\n')

points = torch.tensor([[4.0, 1.0], [5.0, 3.0], [2.0, 1.0]])
points[1][1] # Access individual element
```

Indexing Tensors --> Similar as indexing python lists, with ability to index each dimension

```python
print(points, '\n')
print(points[1:], '\n') # All rows after the first (implicitly all columns)
print(points[1:, :], '\n') # All rows after the first, all columns
print(points[1:, 0], '\n') # All rows after the first, first col
print(points[None], '\n') # Adds a dimention of size 1, just like unsqueeze
```

Naming the Tensors/Indicies of Tensors -- *Not Commonly Used:* \
--> Note: If we want to use tensors outside functions that operate on named tensors, we need to
drop the names by renaming them to None.

```python
weights_named = torch.tensor([0.4, 0.72, 0.08], names = ['channels'])
print(weights_named, '\n')

# Creating the (named) tensor to store image pixel values
img_t = torch.randn(3, 5, 5)
print(img_t.shape)
img_named = img_t.refine_names(..., 'channels', 'rows', 'columns')
print(img_named.shape, img_named.names, '\n')

# "align_as" returns tensor with missing dimensions (for operations with other tensors)
weights_aligned = weights_named.align_as(img_named)
print(weights_aligned)

# Can also specify pass names to operations such as sum
gray_named = (img_named * weights_aligned).sum('channels')
gray_named.shape, gray_named.names
```

#### Data Types (dtype)

```python
double_points = torch.ones(10, 2, dtype=torch.double) # Float-64 tensor
short_points = torch.tensor([[1, 2], [3, 4]], dtype=torch.short) # Int-16 tensor
print(double_points.dtype)
print(short_points.dtype)

# Casting tensor to desired output
double_points = torch.ones(10, 2).to(torch.double)
print(double_points.dtype)

# When mixing input types in operations,  inputs are converted to the larger type
points_64 = torch.rand(5, dtype=torch.double) # 5 Tensor elements with random value b/w 0 and 1
points_short = points_64.to(torch.short)
points_64 * points_short 
```

#### (Some) Tensor Operations

```python
# Operations can be called on tensors
a = torch.ones(3, 2)
a_t = torch.transpose(a, 0, 1)
print(a)
print(a_t)

# And also as a method of the tensor
a_t = a.transpose(0, 1)
```

Tensor are diffrent views of the storage array: Just a C-like array

```python
# How tensor data is actually stored
points = torch.tensor([[4.0, 1.0], [5.0, 3.0], [2.0, 1.0]])
print(points.storage(), '\n')

# Manually indexing storage
points_storage = points.storage()
print(points_storage[0], '\n')
points_storage[1] = 7
print(points.storage(), '\n')

points
```

Any operation/method with a trailing _ (ex. zero_) modifies the input (source) tensor
Any operation/method without a trailing _  leaves the source tensor unchanged and returns a new tensor

```python
a = torch.ones(3, 2)
print(a, '\n')
a.zero_()
print(a)
```

#### Tensor Metadata
Size: A tuple that reps. the shape of tensor (rows, cols, ...) \
Offset: where the tensor start in storage. \
Stride: Tupple representin how many elements to skip over for the next value in (col, row) (1, 3) \
--> Every next element goes in col, every 3rd element starts on new row

```python
second_point = points[1]
print(second_point, '\n')
print(second_point.storage(), '\n') #Notice this is the same storage as the original tensor


# Viewing Sub-tensor metadata
print(second_point.size())
print(second_point.storage_offset())
print(second_point.stride(), '\n')

#Same info as .shape method
print(second_point.shape, '\n')

points.stride()

# Changing the subtensor affects the o.g. tensor. To avoid this, We can clone tensors
second_point[1] = 10.0
```

Changing the subtensor affects the og tensor. To avoid this, We can clone subtenors into new tensors

```python
second_point[1] = 10.0
print(points, '\n')

points = torch.tensor([[4.0, 1.0], [5.0, 3.0], [2.0, 1.0]])

# Clone Tensor
second_point = points[1].clone()
second_point[1] = 10.0
print(points, '\n')

second_point.storage() #Show data of new tensor --> New storage array
```

#### Transposing Tensors

Transposing points in 2d = Swapping strides

```python
print(points)
print(points.size(), '\n')

points_t = points.t() # .t() is shorthand for transpose for 2d tensors
print(points_t)
print(points_t.size(), '\n')

# verify that the two tensors share the same storage
print(id(points.storage()) == id(points_t.storage()), '\n')
print(points[0])
print(points_t[0], '\n')

print(points.stride())
print(points_t.stride())
```

```python
Transposing Mulidimentional arrays
```

```python
some_t = torch.ones(3, 4, 5)
print(some_t)
print(some_t.shape) # [3, 4, 5] = [Dimentions, rows, cols)
print(f'stride: {some_t.stride()} \n')


transpose_t = some_t.transpose(0, 2) #Swap Dim w. cols
print(transpose_t)
print(transpose_t.shape)
print(f'stride: {transpose_t.stride()} \n')

```

#### Contiguous Tensors

```python
print(points.is_contiguous())
print(points_t.is_contiguous(), '\n')

print(points_t)
print(f'stride: {points_t.stride()} \n')

points_t_cont = points_t.contiguous() #Making a new contiguous tensor from non-contiguous: Same data, but diff. storage and stride
print(points_t_cont)
print(f'stride: {points_t_cont.stride()} \n')
```

#### Creating Tensors on GPU

```python
# Can create tensors on GPU by specificing device to contructor
points_gpu = torch.tensor([[4.0, 1.0], [5.0, 3.0], [2.0, 1.0]], device='cuda')

# Or move tensor creted on cpu to the gpu
points_gpu = points.to(device='cuda')

points_gpu = 2 * points_gpu
print(points_gpu)

# move tensor back to cpu
points_cpu = points_gpu.to(device='cpu')
print(points_cpu)
```

Converting b/w tensors and numpy arrays --> Gives access to all python math and science libs that work with np.
- Both share the same underlying storage, so affecting one, effects the other
- np only works on cpu, so if tensor is on gpu, a copy will be made onto the cpu

**Note np arrays default data type is float-64, so we must convert it to float-32 after converting array to tensor** \
--> Faster to work with float-32 and you don't really lose accuracy in model 

```python
points = torch.ones(3, 4)
# From tensor to np array
points_np = points.numpy()
points_np

# From np array to tensor
points = torch.from_numpy(points_np)
points
```

### Serializing (Saving) Tensors

Saving for use with pytorch \
--> .t serialises tensor (for a torchscript model) --> Can be loaded and run independantly (or with model) \
--> .pth is used for saving model weights (into Python Dictionary) --> Needs Original Model

```python
# Save points tensor to ("ourpoints.t") file
torch.save(points, 'data/ourpoints.t') #.save(tensor, path)

# # or using file descriptor
# with open('data/ourpoints.t', 'wb') as f:
#     torch.save(points, f)

#Loading points tensor form ("ourpoints.t") file
torch.load('data/ourpoints.t')
```

Saving Tensor interoperably: Saving in a way that's usable in other software \
Save by converting tensor to np array, then passing it h5py to serialise to HDF5 format

```python
import h5py

f = h5py.File('data/ourpoints.hdf5', 'w')
dset = f.create_dataset('coords', data=points.numpy())
f.close()
```

Indexing data on disk with hdf5

```python
f = h5py.File('data/ourpoints.hdf5', 'r')
dset = f['coords']
last_points = dset[-2:]
print(last_points, '\n')

# Straight from disk to tensor
last_points = torch.from_numpy(dset[-2:])
f.close()
print(last_points)
```
