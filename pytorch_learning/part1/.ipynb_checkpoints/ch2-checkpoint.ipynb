#### We can call alexnet like a function and pass in some parameters
- However, since the network is uninitialized (weights are not loaded) it will return garbage 

```python
# Instantiate the ResNet model and load its weights
resnet = models.resnet101(pretrained=True)
```

```python
#Prints properties and structue of network
# resnet
```

#### Before we can run a pretained model, we need to process our inputs to ***Match what was presented to net during training*** 

```python
from torchvision import transforms
preprocess = transforms.Compose([
    transforms.Resize(256),  #Scale the image to 256 x 256
    transforms.CenterCrop(224), #Crop the image (to 224 x224) around the center
    transforms.ToTensor(), #Transform image to tensor
    transforms.Normalize(  # Normalise RGB
        mean=[0.485, 0.456, 0.406],
        std=[0.229, 0.224, 0.225]
    )])
```

```python
#Open image from files as img
from PIL import Image
img = Image.open("bobby.jpg")
# img.show()
```

```python
# Preporcess the image using fuct. defined above
img_t = preprocess(img)
```

```python
#Reshape tensor to how network expects 
import torch
batch_t = torch.unsqueeze(img_t, 0)
```

#### Now, we can run the pretained model on new data: Called Inferance

```python
#Put model in evaluate mode for inference
resnet.eval()
```

```python
# Passes data through the model and returns the predicted score for each of the 1000 labels 
out = resnet(batch_t)
out
```

```python
# Load Textfile listing labels
with open('dlwpt/data/p1ch2/imagenet_classes.txt') as f:
    labels = [line.strip() for line in f.readlines()]
```

#### Get model prediction: *Index with maximum score in the out tensor*

```python
_, index = torch.max(out, 1) # Index is a one element (1D) tensor, so we need to get the 0th element to get number for coresponmding clabel
percentage = torch.nn.functional.softmax(out, dim=1)[0] * 100 # Model Confidance
labels[index[0]], percentage[index[0]].item()
```

#### Can also fine model seccond, third, ... prediction by soting the score list

```python
_, indecies = torch.sort(out, descending=True)
[(labels[idx], percentage[idx].item()) for idx in indecies[0][:5]]
```
